```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
from sqlalchemy import create_engine
```

```python
engine = create_engine("postgresql://postgres:maga@localhost:5432/contoso_100k")
```

```python
query_2019 = """WITH params AS (
    SELECT 2019 AS target_year
)

SELECT
    COUNT(DISTINCT c.customerkey) AS total_existing_customers,
    SUM(c.total_net_revenue) AS existing_customers_revenue,
    EXTRACT(MONTH FROM c.orderdate) AS month,
    c.cohort_year

FROM cohort_analysis c
CROSS JOIN params p

WHERE
    c.first_purchase_date < MAKE_DATE(p.target_year, 1, 1)
    AND c.orderdate >= MAKE_DATE(p.target_year, 1, 1)
    AND c.orderdate < MAKE_DATE(p.target_year + 1, 1, 1)

GROUP BY month, c.cohort_year

ORDER BY c.cohort_year, month;"""
```

```python
df_2019 = pd.read_sql_query(query_2019, engine)
df_2019['cohort_year'] = df_2019['cohort_year'].astype(int)
```

```python
sns.set_theme(style="white")

# Nombres de los meses
month_names = {
    1: "Jan",
    2: "Feb",
    3: "Mar",
    4: "Apr",
    5: "May",
    6: "Jun",
    7: "Jul",
    8: "Aug",
    9: "Sep",
    10: "Oct",
    11: "Nov",
    12: "Dec"
}

# Copia del DataFrame
df_plot = df_2019.copy()

# Convertir número de mes a nombre
df_plot["month_name"] = df_plot["month"].map(month_names)

# Orden correcto de los meses
month_order = list(month_names.values())

plt.figure(figsize=(10, 6))

sns.lineplot(
    data=df_plot,
    x="month_name",
    y="total_existing_customers",
    hue="cohort_year",
    marker="o",
    linewidth=2,
    palette="tab10"
)

plt.title("Existing Customers by Cohort – 2019")
plt.xlabel("")
plt.ylabel("Number of Existing Customers")

plt.xticks(
    ticks=range(12),
    labels=month_order
)

plt.ylim(0, 250)
plt.grid(False)

plt.legend(
    title="Cohort Year",
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)

plt.tight_layout()
plt.show()
```
